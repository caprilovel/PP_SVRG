```python


def plot_single_result(df1, df2, legend1, legend2, temp, column="train_acc", 
                       log_scale=False, start=0, end=150, save_path=None):
    """
    绘制两个实验结果在同一张图上，并可选择保存为 PDF。
    
    参数说明：
    - df1, df2: 两个DataFrame，分别对应两个实验结果
    - legend1, legend2: 图例名称
    - temp: 温度参数，用于图标题
    - column: 需要绘制的列名（如 'train_acc'）
    - title_prefix: 图标题前缀
    - log_scale: 如果为True，对y轴使用对数刻度（适用于loss）
    - start, end: x轴（epoch）范围
    - save_path: 如果提供路径，则保存为PDF（如 "output.pdf"）
    """
    
    plt.figure(figsize=(6, 4))
    x_vals = range(start, end)
    y1 = df1[column][start:end]
    y2 = df2[column][start:end]

    if log_scale and column == "train_loss":
        plt.semilogy(x_vals, y1, label=legend1, color='blue', linewidth=2)
        plt.semilogy(x_vals, y2, label=legend2, color='red', linewidth=2)
        plt.ylabel(f"{column} (log scale)", fontsize=14)

    else:
        plt.plot(x_vals, y1, label=legend1, color='blue', linewidth=2)
        plt.plot(x_vals, y2, label=legend2, color='red', linewidth=2)
        if column == "grads":
            plt.ylabel(r"$\|\nabla \mathcal{J}(\tilde{\boldsymbol{\theta}}^s, \tilde{\boldsymbol{\theta}}^s)\|^2$", fontsize=14)
        else:
            plt.ylabel(column, fontsize=14)

    plt.xlabel("Epoch", fontsize=14)
    # if end == 40:
    #     plt.xticks(np.arange(start//10 * 10+10, end + 1, 10), fontsize=12)
    # elif end == 400:
    #     plt.xticks(np.arange(start//100 * 100 + 100, end + 1, 100), fontsize=12)
    plt.tick_params(axis='both', which='major', labelsize=12, width=2, length=6)
    plt.tick_params(axis='both', which='minor', width=1.5, length=4)
    plt.gca().spines['top'].set_linewidth(2)
    plt.gca().spines['right'].set_linewidth(2)
    plt.gca().spines['left'].set_linewidth(2)
    plt.gca().spines['bottom'].set_linewidth(2)
    plt.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.7)  # Add grid
    if column == "train_acc":
        plt.legend(loc='lower right', fontsize=24)
    else:
        plt.legend(loc='upper right', fontsize=24)
    plt.tight_layout()
    
    if save_path:
        save_path += f"{temp}_{column}.pdf"
        plt.savefig(save_path, format='pdf')
        print(f"Plot saved to {save_path}")
    
    plt.show()

m = 0
for i in range(3):
    for j in range(3):
        for k in range(3):
            print(f"{names[i]}_{temps[3 * i + j]}_{columns[j]}")
            plot_single_result(
                df1=dfs[3 * i + j][0],
                df2=dfs[3 * i + j][1],
                legend1=legends[0],
                legend2=legends[1],
                temp=temps[3 * i + j],
                column=columns[k],
                log_scale=True,
                start=starts[i],
                end=ends[i],
                save_path=f"plots/{names[i]}"
            )
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# 你可以在这里改成你的 outputs 路径
ROOT = "outputs"

# 要绘制的 metric 名字（需要与你的 CSV header 一致）
METRICS_TO_PLOT = ["train_loss", "train_acc", "eval_loss", "eval_acc"]

# 是否每个 metric 画在独立图上
SEPARATE_FIG = True

# 设定要展示的 epoch 数量
MAX_EPOCH = 80

def clean_exp_name(name):
    """
    去掉类似 20251116-085058_ 的 timestamp 前缀
    """
    parts = name.split("_", 1)
    if len(parts) == 2 and parts[0].replace("-", "").isdigit():
        return parts[1]
    return name


def load_experiment_logs(root, contains='None'):
    """
    扫描 outputs/*/train_stats.csv
    返回 { experiment_name : dataframe }
    """
    logs = {}
    for exp in os.listdir(root):
        if contains != 'None' and contains not in exp:
            continue
        exp_dir = os.path.join(root, exp)
        csv_path = os.path.join(exp_dir, "train_stats.csv")
        if os.path.isfile(csv_path):
            try:
                df = pd.read_csv(csv_path)
                logs[clean_exp_name(exp)] = df
            except Exception as e:
                print(f"[Warning] Failed to read {csv_path}: {e}")

    return logs

```

```python
def plot_metrics(logs, metrics):
    """
    适配你给的 plot_single_result 风格：
    - 边框粗
    - 大字体
    - 蓝/红两色
    - sgd => 虚线
    - svrg => 实线
    """

    def style_from_name(exp_name):
        """根据实验名返回 (color, linestyle)"""
        name = exp_name.lower()
        if "svrg" in name:
            ls = "-"      # 实线
            color = "red" # 红色
        elif "sgd" in name:
            ls = "--"     # 虚线
            color = "blue" # 蓝色
        else:
            # fallback（保持两色循环）
            ls = "-" if "1" in name else "--"
            color = "red" if "1" in name else "blue"
        return color, ls
    def style_from_legend(exp_name):
        """ rewrite the legend, only contains svrg/sgd and lr"""
        name = exp_name.lower()
        rename = ""
        if "svrg" in name:
            rename = "SVRG"
        elif "sgd" in name:
            rename = "SGD"

        lr_part = ""
        for part in name.split("_"):
            if part.startswith("lr"):
                lr_part = part
                break
        if lr_part:
            rename += f"_{lr_part}"
        return rename
        

    if SEPARATE_FIG:
        for metric in metrics:

            plt.figure(figsize=(7, 5))
            plotted_any = False

            for exp, df in logs.items():
                if metric not in df.columns:
                    continue

                df_plot = df.iloc[:MAX_EPOCH]
                if df_plot[metric].isna().all():
                    continue

                color, linestyle = style_from_name(exp)

                plt.plot(
                    df_plot["epoch"],
                    df_plot[metric],
                    label=exp,
                    linewidth=2.0,
                    linestyle=linestyle,
                    color=color,
                )
                plotted_any = True

            if not plotted_any:
                plt.close()
                continue

            # ---- 样式：完全复刻你的 plot_single_result ----
            plt.xlabel("Epoch", fontsize=14)
            plt.ylabel(metric, fontsize=14)

            plt.tick_params(axis='both', which='major', labelsize=12, width=2, length=6)
            plt.tick_params(axis='both', which='minor', width=1.5, length=4)

            ax = plt.gca()
            ax.spines['top'].set_linewidth(2)
            ax.spines['right'].set_linewidth(2)
            ax.spines['left'].set_linewidth(2)
            ax.spines['bottom'].set_linewidth(2)

            plt.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.7)

            plt.legend(loc='upper right', fontsize=12)
            plt.tight_layout()

            plt.savefig(f"plot_{metric}.png", dpi=220)
            print(f"[Saved] plot_{metric}.png")

    else:
        plt.figure(figsize=(7, 5))
        plotted_any = False

        for exp, df in logs.items():
            for metric in metrics:
                if metric not in df.columns:
                    continue

                df_plot = df.iloc[:MAX_EPOCH]
                if df_plot[metric].isna().all():
                    continue

                color, linestyle = style_from_name(exp)

                plt.plot(
                    df_plot["epoch"],
                    df_plot[metric],
                    label=f"{exp}-{metric}",
                    linewidth=2.0,
                    linestyle=linestyle,
                    color=color,
                )
                plotted_any = True

        if not plotted_any:
            plt.close()
            return

        # ---- 统一风格 ----
        plt.xlabel("Epoch", fontsize=14)
        plt.ylabel("Value", fontsize=14)

        plt.tick_params(axis='both', which='major', labelsize=12, width=2, length=6)
        plt.tick_params(axis='both', which='minor', width=1.5, length=4)

        ax = plt.gca()
        ax.spines['top'].set_linewidth(2)
        ax.spines['right'].set_linewidth(2)
        ax.spines['left'].set_linewidth(2)
        ax.spines['bottom'].set_linewidth(2)

        plt.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.7)

        plt.legend(loc='upper right', fontsize=12)
        plt.tight_layout()

        plt.savefig(f"metrics_all.png", dpi=220)
        print(f"[Saved] metrics_all.png")
```
